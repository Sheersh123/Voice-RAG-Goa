```python
import os
import glob
import pickle
import faiss
import pyarrow.parquet as pq
from sentence_transformers import SentenceTransformer

# -----------------------------
# Configuration
# -----------------------------
BASE = "/root/.cache/huggingface/hub/datasets--ai4bharat--MSMARCO-XI/snapshots"
SHARDS = [
    "tamtrain.parquet",
    "hintrain.parquet",
    "bentrain.parquet",
    "gujtrain.parquet",
    "urdtrain.parquet"
]

TARGET_PER_SHARD = 10000
BATCH_SIZE = 512

# -----------------------------
# Load embedding model
# -----------------------------
print("Loading embedding model...")
model = SentenceTransformer("BAAI/bge-small-en-v1.5")

index = None
texts = []
metadata = []

# -----------------------------
# Process each shard
# -----------------------------
for shard_name in SHARDS:

    shard_path = glob.glob(f"{BASE}/*/train/{shard_name}")[0]
    pf = pq.ParquetFile(shard_path)

    language = shard_name[:3]

    print(f"\nProcessing {shard_name}")

    collected = 0

    for batch in pf.iter_batches(batch_size=BATCH_SIZE):

        table = batch.to_pydict()

        answers = table["Answer"]
        queries = table["query"]
        query_ids = table["query_id"]

        remaining = TARGET_PER_SHARD - collected

        if remaining <= 0:
            break

        answers = answers[:remaining]
        queries = queries[:remaining]
        query_ids = query_ids[:remaining]

        embeddings = model.encode(
            answers,
            convert_to_numpy=True,
            show_progress_bar=False
        )

        if index is None:
            index = faiss.IndexFlatL2(embeddings.shape[1])

        index.add(embeddings)

        texts.extend(answers)

        metadata.extend([
            {
                "query": q,
                "query_id": qid,
                "language": language,
                "source": "MSMARCO-XI",
                "chunk_type": "raw"
            }
            for q, qid in zip(queries, query_ids)
        ])

        collected += len(answers)

        if collected % 2000 == 0:
            print(f"  {collected}/{TARGET_PER_SHARD}")

    print(f"Finished {shard_name}: {collected}")

print("\nTotal indexed:", len(texts))
```

```python
from google.colab import drive
import os
import pickle
import faiss

drive.mount('/content/drive', force_remount=False)

SAVE_DIR = "/content/drive/MyDrive/Voice-RAG-Goa/index"
os.makedirs(SAVE_DIR, exist_ok=True)

faiss.write_index(index, f"{SAVE_DIR}/msmarco.index")

with open(f"{SAVE_DIR}/texts.pkl", "wb") as f:
    pickle.dump(texts, f)

with open(f"{SAVE_DIR}/metadata.pkl", "wb") as f:
    pickle.dump(metadata, f)

print("Saved successfully!")
print("Index:", f"{SAVE_DIR}/msmarco.index")
print("Texts:", f"{SAVE_DIR}/texts.pkl")
print("Metadata:", f"{SAVE_DIR}/metadata.pkl")
```
